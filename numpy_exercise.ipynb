**Exercise 1: Create a 1D NumPy array of numbers from 0 to 9**

```python
import numpy as np

array=np.arange(0,10)
print(array)
```

**Exercise 2: Write a code to convert a 1D array to a 2D array with 2 rows.**

```python
nd_array=np.reshape(array,(2,5))
print(nd_array)
```

Exercise 3: Print Array Attributes Instructions: Print the following attributes of the array:

The shape of the array.
The number of array dimensions.
The size of each element in bytes

```python
#Shape of the array

print(nd_array.shape)

#Dimensions of the array
print(nd_array.ndim)

#Size of the array
print(nd_array.size)
```
