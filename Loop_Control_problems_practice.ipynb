**1. Write a program to print all natural numbers from 1 to n using while loop**

> Add blockquote



```python
#using while loop
num=int(input("Enter a number upto which we want to print natural numbers:"))
i=1
print("Printing Natural Nubers using while loop:")
while(i<=num):
  print(i, end=" ")
  i+=1

print("\nNatural numbers printing using for loop")
for i in range(1, num+1):
  print(i,end=" ")
```

**2. Write a program to print all natural numbers in reverse from n to 1 using while loop**

```python

num=int(input("Enter a number to find natural numbers:"))

print("Printing Natural Numbers using while loop in reverse order:")
i=num
while(i>=1):
  print(i,end=" ")
  i-=1

print("\n Printing Natural Numbers using for loop in reverse order:")
for i in range(num,0,-1):
  print(i,end=" ")
```

**3. Write a program to print all alphabets using while loop**

```python
print("Printing alphabets from a-z using while loop")
i=0
while(i<26):
  ch=chr(97+i)
  print(ch,end=" ")
  i+=1

print("\n Printing alphabets from a-z using for loop")
for i in range(0,26,1):
  ch=chr(97+i)
  print(ch,end=" ")
```

**4. Write a program to print all even numbers between 1 to 100 using while loop**

```python
print("Printing even numbers from 1-100 using while loop")
i=1
while(i<=100):
  if(i%2==0):
    print(i,end=" ")
  i+=1


print("\n Printing even numbers from 1-100 using for loop")
lst=[i for i in range(1,101) if i%2==0]
print(lst)
```

**5. Write a program to print all odd numbers between 1 to 100 using while loop**

```python
print("Printing odd numbers from 1-100 using while loop")
i=1
while(i<=100):
  if i%2!=0:
    print(i,end=" ")
  i+=1

print("\nPrinting odd numbers from 1-100 using for loop")
lst=[i for i in range(1,101) if i%2!=0]
print(lst)
```

**6. Write a program to print sum of all natural numbers between 1 to n**

```python
num=int(input("Enter a number:"))
#Using formula
sum_=num*(num+1)/2
print(f"Sum of natural numbers upto {num} is : {sum_}")

#using for loop
sum_=0
for i in range(num+1):
  sum_+=i
print(f"Sum of natural numbers upto {num} is : {sum_}")

#using while loop
i=1
sum_=0
while(i<=num):
  sum_+=i
  i+=1
print(f"Sum of natural numbers upto {num} is : {sum_}")
```

**17. Write a program to check whether a number is palindrome or not**

```python
num=int(input("Enter a number to check whether its palindrom or not:"))

if str(abs(num))==str(abs(num))[::-1]:
  print("Number is palindrome")
else:
  print("Number is not palindrome")
```

**18. Write a program to find frequency of each digit in a given number:**

```python
num=int(input("Enter a number to find freq of its digits: "))

digits=list(map(int,str(abs(num))))

unique_digits=set(digits)

for digit in unique_digits:
  print(f"Frequency of digit {digit} is : {digits.count(digit)}")
```

**21. Write a python program t calculate factorial of number**

```python
import math
num=int(input("Enter a number: "))
print(f"Factorial of number {num} with inbuild function : {math.factorial(num)}")

print("\n Factorial of number using for loop")
fact=1
for i in range(1,num+1):
  fact=fact*i
print(f"Factorial of number {num} using for loop : {fact}")

print("\n Factorial of number using while loop")
i=1
fact=1
while(i<num+1):
  fact=fact*i
  i=i+1
print(f"Factorial of number {num} using while loop : {fact}")


print("\n Factorial of number user defined function")
def  factorial(n):
  if n==0:
    return 1
  else:
    return n*factorial(n-1)

fact=factorial(num)
print(f"Factorial of number {num} using user defined fun : {fact}")
```
